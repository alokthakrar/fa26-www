```python
# Update the make_stripes function to use the correct vertical stripes logic


def make_stripes(N=32, M=32, orientation='horizontal'):
    """
    Utility function to generate different stripe patterns
    """
    if orientation == 'horizontal':
        return np.tile([0, 0, 1, 1], (N, M // 4))
    elif orientation == 'vertical':
        return np.tile([0, 0, 1, 1], (N, M // 4)).T
    elif orientation == 'diagonal':
        return (np.indices((N, M)).sum(axis=0) // 4) % 2
    elif orientation == 'checkerboard':
        i, j = np.indices((N, M))
        return (i // 4 + j // 4) % 2
    elif orientation == 'dots':
        x = np.zeros((N, M))
        x[::4, ::4] = 1
        return x

# Create images using make_stripes function for different orientations
images = {
    "horizontal_stripes": make_stripes(32, 32, 'horizontal'),
    "vertical_stripes": make_stripes(32, 32, 'vertical'),
    "diagonal_stripes": make_stripes(32, 32, 'diagonal'),
    "checkerboard": make_stripes(32, 32, 'checkerboard'),
    "dots": make_stripes(32, 32, 'dots')
}

# Display original images
fig, axs = plt.subplots(1, len(images), figsize=(14, 3.2))
fig.suptitle("Original", fontsize=16, fontweight='bold', y=1.05)
for ax, (name, img) in zip(axs, images.items()):
    ax.imshow(img, cmap='gray')
    ax.set_title(name, fontsize=12)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
def plot_dtfs_magnitude(X, title):
    """
    Plot magnitude of 2D DTFS coefficients |X[k, l]| centered at (0,0).
    """
    X_shifted = np.fft.fftshift(X)
    Ny, Nx = X.shape
    k = np.arange(-Nx // 2, Nx - Nx // 2)
    l = np.arange(-Ny // 2, Ny - Ny // 2)
    extent = [k[0] - 0.5, k[-1] + 0.5, l[0] - 0.5, l[-1] + 0.5]

    plt.figure(figsize=(6, 5))
    plt.imshow(np.log1p(np.abs(X_shifted)), cmap='magma', origin='lower', extent=extent)
    plt.axhline(0, color='w', ls='--', alpha=0.5)  # Horizontal axis through DC
    plt.axvline(0, color='w', ls='--', alpha=0.5)  # Vertical axis through DC
    plt.colorbar(label=r'$\ln(1 + |X[k,l]|)$')
    plt.title(title)
    plt.xlabel('Horizontal Frequency $k$')
    plt.ylabel('Vertical Frequency $l$')
    plt.tight_layout()
    plt.show()

# Plot the DTFS magnitude for each image
for name, coeffs in dtfs_coeffs.items():
    plot_dtfs_magnitude(coeffs, f'Magnitude of DTFS Coefficients for {name}')
```

## (Just for fun) A Thousand Times Faster! 

The `dtfs` function you wrote does $(NM)^2$ operations: one sum over all $NM$ pixels for each of the $NM$ coefficients. At $32 \times 32$ that is about a million terms, which is why it takes a moment every time you call it.

There is a much faster way to compute exactly the same numbers, called the Fast Fourier Transform. It avoids recomputing the same powers of $e^{i2\pi/N}$ over and over, using the same redundancy you saw with the roots of unity. NumPy provides it as `np.fft.fft2`. Run the cell below to compare.

This is the algorithm every real signal processing library uses underneath!


```python
# PROVIDED - just run this cell.
import time

x = images['checkerboard']
N, M = x.shape

t0 = time.perf_counter()
X_yours = dtfs(x)
t_naive = time.perf_counter() - t0

np.fft.fft2(x)                      # warm-up call, not timed
t0 = time.perf_counter()
for _ in range(10):
    X_fast = np.fft.fft2(x) / (N * M)   # fft2 leaves out the 1/(NM), so we divide
t_fast = (time.perf_counter() - t0) / 10

print(f"your dtfs():   {t_naive*1000:9.1f} ms")
print(f"np.fft.fft2:   {t_fast*1000:9.3f} ms")
print(f"speedup:       {t_naive/t_fast:9.0f}x")
print(f"largest difference between the two results: {np.abs(X_yours - X_fast).max():.2e}")
```

```python
def plot_circle(a, d, e):
    """
    You can use this function to plot circles with parameters a,d,e.
    The parameters are described in the homework pdf. 
    
    """
    is_circle = d**2 + e**2 + 4*a > 0
    assert is_circle, "Not a circle"
    
    XLIM_LO = -1
    XLIM_HI = 3
    YLIM_LO = -2
    YLIM_HI = 2
    X_COUNT = 400
    Y_COUNT = 400
    
    x = np.linspace(XLIM_LO, XLIM_HI, X_COUNT)
    y = np.linspace(YLIM_LO, YLIM_HI, Y_COUNT)
    x, y = np.meshgrid(x, y)
    f = lambda x,y: a*(x**2 + y**2) + d*x + e*y
    
    c1 = plt.contour(x, y, f(x,y), [1], colors='r')
    plt.axis('scaled')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title("Best Fit Circle")
```
